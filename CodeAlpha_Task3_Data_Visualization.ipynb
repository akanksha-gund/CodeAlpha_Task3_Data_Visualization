```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    'Product': ['Laptop', 'Phone', 'Tablet', 'Monitor', 'Keyboard',
                'Mouse', 'Printer', 'Headphones', 'Webcam', 'Speaker'],

    'Category': ['Electronics', 'Electronics', 'Electronics', 'Electronics',
                 'Accessories', 'Accessories', 'Electronics', 'Accessories',
                 'Accessories', 'Accessories'],

    'Sales': [75000, 55000, 30000, 25000, 8000,
              5000, 22000, 12000, 9000, 7000],

    'Profit': [12000, 9000, 5000, 4000, 1800,
               1200, 3500, 2500, 1700, 1400]
}

df = pd.DataFrame(data)

df
```

```python
df.head()
```

```python
plt.figure(figsize=(10,5))

plt.bar(df['Product'], df['Sales'])

plt.xlabel('Product')
plt.ylabel('Sales')
plt.title('Sales by Product')

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.bar(df['Product'], df['Profit'])

plt.xlabel('Product')
plt.ylabel('Profit')
plt.title('Profit by Product')

plt.xticks(rotation=45)

plt.show()
```

```python
category_sales = df.groupby('Category')['Sales'].sum()

plt.figure(figsize=(7,5))

plt.bar(category_sales.index, category_sales.values)

plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.title('Sales by Category')

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df['Sales'], df['Profit'])

plt.xlabel('Sales')
plt.ylabel('Profit')
plt.title('Sales vs Profit')

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df['Sales'], bins=5)

plt.xlabel('Sales')
plt.ylabel('Number of Products')
plt.title('Distribution of Sales')

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df['Profit'], bins=5)

plt.xlabel('Profit')
plt.ylabel('Number of Products')
plt.title('Distribution of Profit')

plt.show()
```

```python
category_profit = df.groupby('Category')['Profit'].sum()

plt.figure(figsize=(7,5))

plt.bar(category_profit.index, category_profit.values)

plt.xlabel('Category')
plt.ylabel('Total Profit')
plt.title('Profit by Category')

plt.show()
```

```python
top_products = df.sort_values('Sales', ascending=False)

plt.figure(figsize=(10,5))

plt.bar(top_products['Product'], top_products['Sales'])

plt.xlabel('Product')
plt.ylabel('Sales')
plt.title('Top Products by Sales')

plt.xticks(rotation=45)

plt.show()
```

```python
print("FINAL INSIGHTS")

print("1. Laptop has the highest sales among all products.")
print("2. Laptop also generates the highest profit.")
print("3. Electronics category has higher total sales than Accessories.")
print("4. Electronics category also has higher total profit than Accessories.")
print("5. Higher sales generally show a positive relationship with profit in this dataset.")
```
